```python
from typing import Annotated
from fastapi import FastAPI,Path,Query

app = FastAPI()

@app.get("/items/{item_id}")
def read_items(
    # 对路径参数进行参数验证
    item_id:Annotated[int,Path(title="The ID of the item to get")],
    q:Annotated[str | None ,Query(alias="item-query")] = None,
):
    results ={"item_id":item_id}
    if q:
        results.update({"q":q})
    return results
```

```python
# 如果将没有默认值的参数放在有默认值参数后，python会报错
# 需要将没有默认值参数放在有默认值参数前
@app.get("/items/{item_id}")
def read_items(q:str,item_id:int = Path(title="The ID of the item to get")):
    results = {"item_id":item_id}
    if q:
        results.update({"q":q})
    return results
```

```python

@app.get("/items/{item_id}")
async def read_items(*, item_id: int = Path(title="The ID of the item to get"),
                     q: str
                     ):
    results = {"item_id": item_id}
    if q:
        results.update({"q": q})
    return results
```

```python
# 大于等于
@app.get("/items/{item_id}")
def read_items(
    item_id:Annotated[int,Path(title='The ID of the item to get',ge=1)],
    q:str
):
    results = {"item_id":item_id}
    if q:
        results.update({"q":q})
    return results
```

```python
# 大于和小于等于

@app.get("/items/{item_id}")
def read_items(
    item_id:Annotated[int,Path(title="The ID of the item to get",gt=0,le=1000)],
    q:str
):
    results={"item_id":item_id}
    if q:
        results.update({"q":q})
    return results
```

```python
# 浮点数、大于和小于

@app.get("/items/{item_id}")
def read_items(
    item_id:Annotated[int ,Path(title='The ID of the item to get',gt=0,le=1000)],
    q:str,
    size:Annotated[float,Query(gt=0,lt=10.5)]
):
    results={"item_id":item_id}
    if q:
        results.update({"q":q})
    if size:
        results.update({"size":size})
    return results
```
