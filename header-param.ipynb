```python
# 
from typing import Annotated
from fastapi import FastAPI ,Header

app = FastAPI()


@app.get("/items/")
# python 会自动将user_agent 转为 user-agent
def read_items(user_agent: Annotated[str | None,Header()] = None):
    return {"User-Agent",user_agent}
```

```python
# 禁用下划线自动转换为连字符

@app.get("/items/")
def read_items(strange_header:Annotated[str | None,Header(convert_inderscores=False)]):
    return {"strange_header":strange_header}
```

```python
# 重复请求头，一个请求头中一个参数有多个值
@app.get("/items/")
def read_items(x_token:Annotated[list[str] | None,Header()] = None):
    return {"X-Token values": x_token}
```
