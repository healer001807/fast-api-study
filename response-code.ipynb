```python
from fastapi import FastAPI,status

 app = FastAPI()
```

```python
@app.get("/items/",status_code =201)
def create_item(name: str):
    return {'name':name }
```

```python
# 使用http.status 
@app.get('/items/',status_code= status.HTTP_201_CREATED)
def create_item(name: str):
    return {'name':name}
```
