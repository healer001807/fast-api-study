```python
from typing import Annotated
from fastapi import FastAPI,Form
from pydantic import BaseModel

app = FastAPI()
```

```python
class FormData(BaseModel):
    username: str
    password: str
    # 禁止额外字段
    model_config = {'extra':'forbid'}

@app.post('/login/')
def login(data: Annotated[FormData,Form()]):
    return data
```
